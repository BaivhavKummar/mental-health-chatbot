```python
# --- Imports ---
import cv2
import threading
import queue
import time
import numpy as np
import sounddevice as sd
import speech_recognition as sr
from deepface import DeepFace
from IPython.display import display, clear_output
import ipywidgets as widgets
import random
import os
import json # For potential future structured data logging

# NLP
from transformers import pipeline

# LLM
import google.generativeai as genai

# TTS
import pyttsx3

print("Libraries loaded.")
# --- Configuration ---
AUDIO_SAMPLE_RATE = 16000
AUDIO_CHANNELS = 1
SPEECH_PHRASE_TIME_LIMIT = 7 # Slightly longer limit
DEEPFACE_DETECTOR_BACKEND = 'opencv'
DEEPFACE_MODEL_NAME = 'VGG-Face'
DETECTION_INTERVAL = 5 # Frames between face detection runs

# --- Shared Resources ---
text_q = queue.Queue()          # Queue for recognized text from STT
tts_q = queue.Queue()           # Queue for text the bot needs to speak
stop_event = threading.Event()  # Signal threads to stop
listen_event = threading.Event()# Signal STT thread to listen
emotion_lock = threading.Lock() # Lock for global emotion variable
chat_history_lock = threading.Lock() # Lock for chat history
last_detected_emotion_global = "N/A"
chat_history = [] # Store conversation context {role: 'user'/'model', parts: [text]}

print("Shared resources initialized.")
```

```python

# --- NLP Model Setup ---
print("Loading NLP models...")
nlp_sentiment_analyzer = None
nlp_emotion_analyzer = None
try:
    # Using pipelines is easiest
    nlp_sentiment_analyzer = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
    # Emotion model (choose one, roberta-base-go_emotions is comprehensive but heavier)
    # nlp_emotion_analyzer = pipeline("text-classification", model="SamLowe/roberta-base-go_emotions", top_k=3)
    nlp_emotion_analyzer = pipeline("text-classification", model="j-hartmann/emotion-english-distilroberta-base", top_k=None) # Simpler one
    print("NLP models loaded successfully.")
except Exception as e:
    print(f"Warning: Could not load NLP models: {e}. Text analysis will be limited.")

def analyze_text_nlp(text):
    """Analyzes text using loaded NLP models."""
    analysis = {'sentiment': None, 'text_emotion': None}
    if nlp_sentiment_analyzer:
        try:
            result = nlp_sentiment_analyzer(text)[0]
            analysis['sentiment'] = result # {'label': 'POSITIVE', 'score': 0.99}
        except Exception as e:
            print(f"NLP Sentiment analysis failed: {e}")
    if nlp_emotion_analyzer:
        try:
            result = nlp_emotion_analyzer(text)[0]
             # Sort results if model returns multiple emotions
            sorted_emotions = sorted(result, key=lambda x: x['score'], reverse=True)
            analysis['text_emotion'] = sorted_emotions # [{label: 'joy', score: 0.8}, ...]
        except Exception as e:
            print(f"NLP Emotion analysis failed: {e}")
    return analysis

# --- LLM (Gemini) Setup ---
print("Configuring LLM...")
llm_model = None
try:
    GOOGLE_API_KEY = os.environ.get('GOOGLE_API_KEY')
    if GOOGLE_API_KEY:
        genai.configure(api_key=GOOGLE_API_KEY)
        llm_model = genai.GenerativeModel('gemini-pro') # Or other compatible model like gemini-1.5-flash
        print("Google Gemini configured successfully.")
    else:
        print("Warning: GOOGLE_API_KEY environment variable not found. LLM features disabled.")
except Exception as e:
    print(f"Error configuring Google Gemini: {e}. LLM features disabled.")


# --- TTS Setup ---
print("Initializing TTS engine...")
tts_engine = None
try:
    tts_engine = pyttsx3.init()
    # Optional: Adjust properties (examples)
    # tts_engine.setProperty('rate', 150) # Speed percent (can go over 100)
    # tts_engine.setProperty('volume', 0.9) # Volume 0-1
    # voices = tts_engine.getProperty('voices')
    # tts_engine.setProperty('voice', voices[1].id) # Change voice if available
    print("TTS engine initialized.")
except Exception as e:
    print(f"Warning: Could not initialize TTS engine: {e}. Speech output disabled.")

def text_to_speech_thread():
    """
    Waits for text on tts_q and speaks it using pyttsx3.
    runAndWait() blocks this thread while speaking.
    """
    print("TTS thread started.")
    while not stop_event.is_set():
        try:
            text_to_speak = tts_q.get(timeout=0.5)
            if tts_engine:
                print(f"[TTS Speaking]: {text_to_speak}")
                tts_engine.say(text_to_speak)
                tts_engine.runAndWait() # Blocks this thread until speaking is done
                print("[TTS Finished Speaking]")
            else:
                 # TTS disabled, just print indication
                 print(f"[TTS Disabled - Would say]: {text_to_speak}")
                 time.sleep(0.5) # Simulate speak time slightly
            tts_q.task_done() # Mark task as completed
        except queue.Empty:
            continue # No text to speak, check stop_event
        except Exception as e:
            print(f"Error in TTS thread: {e}")
            # Make sure task_done is called even if say/runAndWait fails
            if not tts_q.empty():
                 try: tts_q.task_done()
                 except ValueError: pass # Handle race condition if already done


    # Drain queue on exit? Maybe not necessary as runAndWait waits.
    print("TTS thread finished.")


# --- Speech Recognition Thread (Mostly unchanged, adds placeholders on error) ---
r = sr.Recognizer()
def speech_recognition_thread():
    print("Speech recognition thread started.")
    mic = sr.Microphone(sample_rate=AUDIO_SAMPLE_RATE)
    adjustment_duration = 0.8
    with mic as source:
        try:
            print(f"Adjusting for ambient noise ({adjustment_duration}s)...")
            r.adjust_for_ambient_noise(source, duration=adjustment_duration)
            print("Adjustment complete.")
        except Exception as e: print(f"Ambient noise adjustment failed: {e}")

    while not stop_event.is_set():
        if listen_event.wait(timeout=1.0): # Wait for signal
            print("\n>>> Listening... Speak now! <<<")
            audio_data = None
            text = "[No Speech Triggered]" # Default text
            try:
                with mic as source:
                    audio_data = r.listen(source, phrase_time_limit=SPEECH_PHRASE_TIME_LIMIT)
                print(">>> Processing speech... <<<")
                text = r.recognize_google(audio_data)
                text_q.put(text)
            except sr.WaitTimeoutError: text = "[Recognition timeout]"
            except sr.UnknownValueError: text = "[Audio not understood]"
            except sr.RequestError as e: text = f"[API Error: {e}]"
            except Exception as e: text = f"[Recognition System Error: {e}]"
            finally:
                if text.startswith("["): print(text) # Print errors immediately
                text_q.put(text) # Put result (or error placeholder) onto queue
                listen_event.clear()

    print("Speech recognition thread finished.")

# --- Video Handling Thread (Updates global emotion with lock) ---
def video_processing_thread():
    global last_detected_emotion_global
    print("Video processing thread started.")
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("Error: Could not open webcam."); stop_event.set(); return

    print("Webcam opened successfully.")
    last_emotion_local = "N/A"; last_face_box = None
    frame_count = 0

    while not stop_event.is_set():
        ret, frame = cap.read()
        if not ret: time.sleep(0.1); continue

        display_frame = frame.copy(); frame_count += 1

        if frame_count % DETECTION_INTERVAL == 0:
            current_detected_emotion = "N/A"; current_face_box = None
            try:
                analysis_results = DeepFace.analyze(img_path=frame, actions=['emotion'], detector_backend=DEEPFACE_DETECTOR_BACKEND, enforce_detection=False, silent=True)
                if analysis_results and len(analysis_results) > 0:
                     result = analysis_results[0]
                     current_detected_emotion = result['dominant_emotion']
                     facial_area = result['region']
                     current_face_box = (facial_area['x'], facial_area['y'], facial_area['w'], facial_area['h'])
            except Exception as e: current_detected_emotion = "Error" # Log error state

            last_emotion_local = current_detected_emotion
            last_face_box = current_face_box
            with emotion_lock: # Update global safely
                last_detected_emotion_global = current_detected_emotion

        # Drawing logic (same as before)
        if last_face_box:
            x, y, w, h = last_face_box
            cv2.rectangle(display_frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
            text_to_display = f"Emotion: {last_emotion_local}"
            (tw, th), _ = cv2.getTextSize(text_to_display, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 2)
            tx, ty = x, y - 10 if y > 20 else y + h + 20
            cv2.rectangle(display_frame, (tx, ty - th - 5), (tx + tw, ty + 5), (0, 0, 0), -1)
            cv2.putText(display_frame, text_to_display, (tx, ty), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

        cv2.putText(display_frame, "L: Listen | Q: Quit", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 0), 2, cv2.LINE_AA)
        cv2.putText(display_frame, "L: Listen | Q: Quit", (10+1, 30+1), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2, cv2.LINE_AA)
        cv2.imshow('Live Emotion Detection & Chatbot', display_frame)

        key = cv2.waitKey(1) & 0xFF
        if key == ord('q'): stop_event.set(); break
        if key == ord('l') and not listen_event.is_set(): listen_event.set()

    print("Video processing thread finished.")
    cap.release(); cv2.destroyAllWindows()
    for i in range(5): cv2.waitKey(1)


# --- Enhanced Chatbot Logic Function ---
def get_chatbot_reply_enhanced(user_text, facial_emotion):
    """
    Generates a reply using NLP analysis and potentially an LLM.
    Manages chat history.
    """
    global chat_history
    bot_response = "..." # Default response

    # --- Basic Error Handling ---
    if user_text.startswith("[") and user_text.endswith("]"):
        if user_text == "[Audio not understood]": return random.choice(["Sorry, I didn't quite get that.", "Could you say that again?", "My audio input failed."])
        elif user_text == "[Recognition timeout]": return random.choice(["I didn't hear anything.", "Did you say something?", "Listening timed out."])
        else: return "I'm experiencing a technical difficulty with speech input."

    # --- Perform NLP Analysis ---
    text_analysis = analyze_text_nlp(user_text)
    text_sentiment = text_analysis.get('sentiment', {}).get('label', 'UNKNOWN')
    primary_text_emotion = text_analysis.get('text_emotion', [{}])[0].get('label', 'unknown') if text_analysis.get('text_emotion') else 'unknown'
    print(f"[NLP Analysis]: Text Sentiment='{text_sentiment}', Text Emotion='{primary_text_emotion}'")

    # --- Construct Context String for LLM ---
    context_prompt = f"""You are an empathetic AI assistant integrated with real-time emotion detection. Respond kindly and supportively.
    Keep responses relatively concise for a chat interface. Avoid lists unless necessary.
    If the user mentions severe distress, thoughts of self-harm, or ending their life, strongly but gently advise them to contact a crisis hotline (like 988 in the US/Canada, or Befrienders Worldwide internationally) or seek immediate professional help, explaining you are an AI and cannot provide the necessary support for such situations.

    Current user facial emotion detected: {facial_emotion}
    Current user text sentiment: {text_sentiment}
    Primary emotion detected in user text: {primary_text_emotion}

    Conversation History (for context):"""
    # Prepare history (limit length if needed)
    history_limit = 10 # Keep last 5 user/model turns
    limited_history = chat_history[-history_limit:]

    # --- Call LLM (if available) ---
    if llm_model:
        try:
            # Add latest user message to history before sending
            with chat_history_lock:
                 chat_history.append({'role': 'user', 'parts': [user_text]})
                 # Re-prepare limited history including the new message
                 limited_history = chat_history[-history_limit:]

                 # Prepare content for Gemini API
                 gemini_content = []
                 gemini_content.append({'role': 'user', 'parts': [context_prompt]}) # Prepend context as 'user' (adjust if needed based on best practices)
                 gemini_content.append({'role': 'model', 'parts': ["Okay, I understand the context. How can I help the user based on their latest message?"]}) # Prime the model

                 # Add conversation turns
                 for turn in limited_history:
                      role = 'user' if turn['role'] == 'user' else 'model'
                      parts = turn['parts']
                      gemini_content.append({'role': role, 'parts': parts})

            # Make the API call
            response = llm_model.generate_content(gemini_content, stream=False)

            if response.parts:
                bot_response = response.text
            else: # Handle blocked response
                 print("Warning: LLM response blocked or empty. Using fallback.")
                 # Check response.prompt_feedback for block reasons
                 bot_response = "I'm having trouble formulating a response right now. Can you rephrase or try a different topic?"
                 # Fallback to basic rules might be better here
        except Exception as e:
            print(f"LLM Error: {e}")
            bot_response = random.choice([ # Fallback on error
                "My thinking circuits are a bit fuzzy right now. Could you repeat that?",
                "Hmm, I encountered an issue. Let's try again.",
            ])
    else:
        # --- LLM Disabled - Very Basic Rule-Based Fallback ---
        bot_response = get_chatbot_reply(user_text, facial_emotion) # Call the old simple version

    # --- Update Chat History with Bot Response ---
    with chat_history_lock:
        chat_history.append({'role': 'model', 'parts': [bot_response]})
        # Optional: Prune history if it gets too long beyond the limit we send
        if len(chat_history) > 50: chat_history = chat_history[-history_limit:] # Keep last N turns total


    # --- Handle specific exit command ---
    if "goodbye now please quit" in user_text.lower(): # More specific quit phrase
        stop_event.set()
        bot_response = "Okay, shutting down. Take care!"

    return bot_response


# --- Old Simple Chatbot Logic (Used as Fallback if LLM disabled) ---
def get_chatbot_reply(user_text, current_emotion):
     user_text_lower = user_text.lower()
     # ... (Keep the simple keyword logic from the previous step here) ...
     if "hello" in user_text_lower: return "Hi there! How are things?"
     elif "sad" in user_text_lower: return "I hear you're feeling sad. I'm listening."
     # Add more simple rules as needed
     else: return f"Okay, you mentioned '{user_text[:20]}...'. Tell me more." # Generic fallback


# --- Main Execution Loop ---
print("Starting threads...")
video_thread = threading.Thread(target=video_processing_thread, daemon=True) # Daemon True might help cleanup?
stt_thread = threading.Thread(target=speech_recognition_thread, daemon=True)
tts_thread = threading.Thread(target=text_to_speech_thread, daemon=True)

video_thread.start()
stt_thread.start()
if tts_engine: tts_thread.start() # Only start if initialized

output_widget = widgets.Output(layout={'border': '1px solid black', 'height': '250px', 'overflow_y': 'scroll'})
print("\n--- Conversation --- (Appears below)")
display(output_widget)

# Initial Greeting
initial_greeting = "Hello! I'm ready. Press 'L' in the video window and speak."
with output_widget: print(f"Bot: {initial_greeting}")
if tts_engine: tts_q.put(initial_greeting)

try:
    while not stop_event.is_set():
        try:
            recognized_text = text_q.get(timeout=0.5)
            with output_widget: print(f"\nYou: {recognized_text}")

            if recognized_text == "[No Speech Triggered]": continue # Skip chatbot logic if no actual speech

            current_emotion = "N/A"
            with emotion_lock: current_emotion = last_detected_emotion_global

            chatbot_response = get_chatbot_reply_enhanced(recognized_text, current_emotion)

            with output_widget: print(f"Bot: {chatbot_response}")
            if tts_engine: tts_q.put(chatbot_response) # Queue the response for speaking

        except queue.Empty:
            # Check if critical threads died unexpectedly
            if not video_thread.is_alive() or not stt_thread.is_alive():
                if not stop_event.is_set(): print("Error: A critical thread stopped. Shutting down.")
                stop_event.set()
            continue
        except Exception as e:
             print(f"\n*** Error in main loop: {e} ***\n")
             time.sleep(0.1)

except KeyboardInterrupt: print("\nKeyboard Interrupt. Stopping threads...")
finally:
    stop_event.set()

    print("Waiting for threads to finish...")
    # Give threads a moment to potentially finish their current task
    video_thread.join(timeout=2.0)
    stt_thread.join(timeout=2.0)
    if tts_engine:
        # Wait for TTS queue to be processed before exiting
        tts_q.join() # Wait for all items to be processed by tts_thread
        tts_thread.join(timeout=5.0) # Allow time for final runAndWait()

    cv2.destroyAllWindows()
    for i in range(5): cv2.waitKey(1)

    print("\nApplication finished.")
```
